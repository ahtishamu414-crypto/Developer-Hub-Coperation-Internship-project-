```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as mlt
import seaborn as sns
```

```python
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

# SECTION 1 — LOAD DATASET

```python
iris=pd.read_csv("iris.csv")
```

# SECTION 2 — DATASET INSPECTION


```python
print(f"\n📐 Shape         : {iris.shape}")
print(f"📋 Columns       : {list(iris.columns)}")

print("\n🔍 First 5 Rows")
print(iris.head().to_string())
```

#SECTION 3 — STYLE CONFIGURATION

## Scatter Plot-to analyze relationships between variables.

```python
sns.scatterplot(
    data=iris,
    x="SepalLengthCm",
    y="SepalWidthCm",
    hue="Species"
)

plt.title("Sepal Length vs Sepal Width (Iris Dataset)")
plt.show()
```

## Histogram-to examine data distribution.

```python
iris.hist(color="skyblue", edgecolor="black", figsize=(12, 8), bins=15)
plt.tight_layout()
plt.show()
```
